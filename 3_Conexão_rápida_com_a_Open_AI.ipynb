## 3.1 Sobre a OpenAI
----

### 3.1.1 O que é a OpenAI?
> A OpenAI é uma empresa e laboratório de pesquisa de inteligência artificial (fundada em 2015) que desenvolve modelos de IA de texto, áudio, imagem, vídeo e raciocínio. Ela possui dois produtos, o Chat GPT e API.

- **ChatGPT:** é o aplicativo e site de conversa voltado ao usuário final, com interface visual pronta para tirar dúvidas, escrever textos, analisar arquivos e conversar. Possui planos próprios (como gratuito e pagos)

- **API da OpenAI:** é o serviço voltado a desenvolvedores e empresas para integrar os modelos da OpenAI dentro de outros sistemas, sites, robôs ou aplicativos via código. A cobrança é separada do ChatGPT e funciona, em geral, por consumo de tokens (pagamento conforme o uso).

#### 3.2.2.1 Importação de ferramentas - Célula 1


```python
# Célula 1 - imports

import os
from dotenv import load_dotenv
from openai import OpenAI
```

<br>

#### 3.2.2.2 Carregamento e validação da chave - Célula 2


```python
# Célula 2 - Carregue as variáveis de ambiente em um arquivo chamado .env.

load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("API Key não encontrada.")

   
client = OpenAI(api_key=api_key)

print("API Key encontrada com sucesso.")
```

<br>

#### 3.2.2.3 Configuração de chamada - Célula 3
> O que faz?
Cria a estrutura de mensagens que será enviada para a API.
>
> Explicação:
> Na API da OpenAI, as mensagens não são enviadas como um texto solto, mas sim estruturadas em uma lista de dicionários onde cada item possui uma role (papel). Isso diz ao modelo quem está falando e qual é o contexto da instrução:
>
> Cito um exemplo opcional logo abaixo da Célula3

**1. system:** Define a "personalidade", o contexto, as restrições ou as regras de comportamento da IA. (Ex: "Você é um programador sênior muito paciente..."). Geralmente é o primeiro item da lista.

**2. user:** É a entrada enviada por você (a sua pergunta, comando ou dúvida).

**3. assistant:** São as respostas geradas pelo próprio modelo. Em conversas contínuas (chat com histórico), as respostas anteriores da IA entram com essa role para que ela lembre do que foi dito.


```python
# Célula 3 - Configuração de chamada

# role define quem enviou a mensagem:
# system, user ou assistant

mensagens = [
    {
        "role": "system",
        "content": "Você é um assistente especialista em Humor, sempre contando uma piada no final."
    },
    {
        "role": "user",
        "content": "Olá, GPT! Esta é a minha primeira mensagem para você via API! Pode me dar boas-vindas?"
    }
]

mensagens
```

<br>

#### 3.2.2.4 Acionando a OpenAI, e configurando o método de receber respostas - Célula 4


```python
# Célula 4 - Acionando a OpenAI, e configurando o método de receber respostas 


try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=mensagens
    )

    print(response.choices[0].message.content)

except Exception as e:
    print(f"Erro: {e}")
```

## 3.3 Segurança

----
### 3.3.1 Segurança no GitHub: O perigo de expor sua API Key

>Chaves de API funcionam como senhas mestras. Se você subir o seu arquivo .env contendo a sua chave real para um repositório público no GitHub, robôs automatizados podem encontrá-la em segundos e usá-la para consumir créditos da sua conta. <br><br> Para evitar isso, adotamos duas práticas fundamentais no desenvolvimento profissional:

<br>

#### 3.3.1.1 O arquivo .gitignore
 
> O arquivo .gitignore avisa ao Git quais arquivos e pastas **nunca devem ser enviados para o repositório remoto (GitHub).**

- Crie um arquivo chamado .gitignore na raiz do seu projeto, dentro dele digite:

> `.env`
 
 - Dessa forma, o Git vai ignorar completamente o seu arquivo com a chave real, mantendo-o apenas na sua máquina local.

<br>

#### 3.3.1.2 Arquivo .env.example (Documentação do Projeto)
>Como outras pessoas (ou você mesmo em outro computador) precisarão saber quais variáveis de ambiente o projeto exige para funcionar, criamos um arquivo de modelo chamado .env.example.

- Crie um arquivo chamado .env.example na raiz do projeto.

- Deixe o conteúdo dele sem valores reais, apenas indicando a estrutura necessária:
  >`OPENAI_API_KEY=`sua_chave_aqui_sk-proj-...

Este arquivo .env.example pode e deve ser enviado para o GitHub. Quando alguém clonar o seu repositório, basta copiar o .env.example, renomeá-lo para .env e preencher com a própria chave.

## 3.4 Monitorando Custos e Tokens
----

### 3.4.1 O que são tokens? 
> Tokens são as unidades básicas que os modelos da OpenAI usam para ler, processar e gerar texto. Um token pode ser uma palavra inteira, parte de uma palavra, um caractere, um espaço ou um sinal de pontuação.
> 
> Como vimos na introdução, a cobrança da API da OpenAI é feita por token (pedaços de palavras processadas). Então a cobrança é feita por requisição podendo váriar.
>
> Como exemplo para textos, em inglês, 1 token equivale a cerca de 4 caracteres ou ¾ de uma palavra (100 tokens ≈ 75 palavras). Em português, a proporção costuma ser maior (cerca de 1,5 a 1,8 token por palavra, dependendo do texto e do modelo)

<br>

### 3.4.2 Controle de token via OpenAi
> Na plataforma da OpenAI, com usuário logado, basta ir em custos e colocar um valor limite, ao atingir esse limite as requisições são bloqueadas.

<br>

### 3.4.3 Controle de token via python


> 1. prompt_tokens: Tokens consumidos pelo texto que você enviou (sua pergunta + as instruções do sistema).

> 2. completion_tokens: Tokens gerados pela IA na resposta dela.

> 3. total_tokens: A soma dos dois (o que será descontado dos seus créditos na API)

<br>

#### 3.4.3.1 Atualizando a Célula 4 com a Inspeção de Tokens
Substitua o conteúdo da sua Célula 4 para capturar e exibir essas informações junto com a resposta:

```python
# Célula 5 - Acionando a OpenAI, exibindo a resposta e inspecionando o uso de tokens

try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=mensagens
    )

    # Exibindo o texto gerado pela IA
    print("--- Resposta da IA ---")
    print(response.choices[0].message.content)
    
    # Exibindo o relatório de tokens da requisição
    print("\n--- Relatório de Consumo (Tokens) ---")
    print(f"Tokens de Entrada (Prompt): {response.usage.prompt_tokens}")
    print(f"Tokens de Saída (Resposta): {response.usage.completion_tokens}")
    print(f"Total de Tokens Consumidos: {response.usage.total_tokens}")

except Exception as e:
    print(f"Erro: {e}")
```
