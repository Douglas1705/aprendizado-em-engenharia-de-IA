## 2.1 O que são notebooks Jupyter?

> Jupyter Notebook é um ambiente **interativo** que permite combinar **código**, **texto**, **imagens**, **gráficos** e **resultados de execução** em um único documento.
> Os arquivos são .ipynb formatados em JSON, combinando com código executável,
> É muito utilizado em ciência de dados, machine learning, ensino e prototipação.


### 2.1.1 Para que servem?

- Explorar dados rapidamente
- Criar análises reproduzíveis
- Documentar raciocínios junto ao código
- Ensinar programação e conceitos técnicos
- Compartilhar experimentos


### 2.1.2 Exemplos de como usar

#### O que é um Kernel?
> O Kernel é o processo responsável por executar o código do notebook. Sempre que você executa uma célula, a instrução é enviada ao Kernel.
> No cursor ao instalarmos o UV ele cria um ambiente virtual utilizando o python 3.12 (2026), na pasta .venv, para utilizarmos o kernel basta clicar no botão selecionar kernel no canto superior esquerdo, e escolher .venv python3.12


#### Como rodar células?
- No cursor temos 4 botões de play, o primeiro a esquerda roda a célula selecionada;
- Shift + Enter: executa e avança para a próxima célula; 
- Ctrl + Enter: executa e permanece na célula atual;


#### Como resetar o notebook?
> Reinicie o Kernel para limpar variáveis e estados salvos em memória. No cursor no campo superior está escrito restart, ele reseta os campos executdos mas deixa os resultados visíveis.


#### Output de saída
> Toda execução gera uma saída, como texto, tabelas, gráficos ou mensagens de erro. No cursor basta clicar em clear all outputs e os resultados somem.
