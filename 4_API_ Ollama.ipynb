## 4.1 O que é Ollama?

> O Ollama é uma ferramenta open-source desenvolvida para facilitar a execução de Grandes Modelos de Linguagem (LLMs) localmente, diretamente na sua própria máquina. Ele abstrai toda a complexidade de configuração de ambiente, gerenciamento de pesos dos modelos e otimização de hardware (CPU e GPU), empacotando tudo em uma interface simples de linha de comando (CLI) e uma API REST local.

> Utilizar modelos localmente com o Ollama oferece vantagens significativas em relação às APIs tradicionais em nuvem:

1. **Privacidade e Segurança:** Os dados e prompts enviados aos modelos nunca saem da sua rede local.

2. **Custo Zero de API:** Não há cobrança por tokens processados, permitindo experimentação ilimitada.

3. **Operação Offline:** Uma vez que o modelo foi baixado, você pode utilizá-lo sem conexão com a internet.

4. **Baixa Latência de Rede:** Elimina o tempo de resposta de servidores externos, dependendo apenas do poder de processamento do seu hardware.

## 4.3 Modelos

> O Ollama conta com uma biblioteca extensa de modelos pré-configurados e prontos para uso, otimizados para diferentes capacidades de hardware (através de técnicas como quantização). 
> 
>Alguns modelos tem opcão gratuita e alguns necessitam de Key pagas.

Entre os mais populares estão:

- Llama 3 (Meta): Excelente equilíbrio entre desempenho geral, raciocínio lógico e compreensão de texto.

- Mistral / Mixtral: Modelos altamente eficientes e rápidos, muito utilizados para tarefas de geração de texto e resumo.

- Gemma (Google): Modelos leves e de código aberto baseados na mesma tecnologia do Gemini.

- CodeLlama / DeepSeek-Coder / Qwen2.5-Coder: Modelos especializados em geração, autocompletar e depuração de código.

- Nomic-embed-text: Modelo focado na geração de embeddings (vetores numéricos), essencial para aplicações de RAG (Retrieval-Augmented Generation).

Você pode explorar todos os modelos disponíveis e suas variações de tamanho (ex: 7B, 8B, 70B de parâmetros) na biblioteca oficial do Ollama em: https://ollama.com/search

### 4.3.1 Sobre modelos

#### 4.3.1.1 Pesos
> Em modelos Ollama encontramos nomes com parametros, ou pesos, os mais comuns em milhões ou bilhões, como 7M ou 27B, pesos aparecem ao lado no nome do modelo, exemplo no caso do modelo gemma:
>
> gemma3:270m, gemma3:1b, gemma3:4b, gemma3:12b, gemma3:27b.
>
> O número de parâmetros representa a quantidade de valores ajustáveis que o modelo aprendeu durante o treinamento.


Pense assim:

- Um modelo de 7 bilhões (7B) parâmetros tem aproximadamente 7 bilhões de "pesos" internos.

> Esses parâmetros armazenam o conhecimento e os padrões que o modelo aprendeu a partir dos dados de treinamento.

Os pesos causam:

- Maior capacidade de raciocínio

- Mais conhecimento armazenado

- Mais consumo de memória RAM

- Maior necessidade de processamento

Observação: Às vezes um modelo moderno de 8B supera um modelo antigo de 13B porque ele foi treinado com dados melhores e técnicas mais recentes.

#### 4.3.1.2 Pesos e mémoria ram
> Cuidao ao realizar download de modelos, verifique a capacidade da sua mémoria ram para não travar o hardware


| Tamanho  | RAM necessária (aprox.) |
| ------------- |:-------------:|
| 7B       | 4 a 8 GB      |
| 13B      | 8 a 12 GB     |
| 32B      | 20 a 30 GB    |
| 70B      | 40+ GB        |

## 4.4 Comandos Úteis do Ollama

A interface de linha de comando (CLI) do Ollama é inspirada no Docker, tornando o gerenciamento de modelos bastante intuitivo. Abaixo estão os comandos essenciais para o dia a dia:

| Comando | Descrição |
|----------|----------|
| `ollama pull <modelo>` | Baixa um modelo da biblioteca oficial para a sua máquina (ex.: `ollama pull llama3`). |
| `ollama run <modelo>` | Inicia uma sessão interativa de chat com o modelo no terminal. Se o modelo não estiver baixado, ele fará o download automaticamente. |
| `ollama list` | Lista todos os modelos armazenados localmente e o espaço em disco que ocupam. |
| `ollama ps` | Mostra quais modelos estão carregados na memória (RAM/VRAM) no momento. |
| `ollama rm <modelo>` | Remove um modelo específico da sua máquina para liberar espaço em disco. |
| `ollama serve` | Inicia o servidor local do Ollama manualmente (caso ele não esteja rodando em segundo plano). |

## Exemplos

```bash
# Baixar um modelo
ollama pull llama3

# Executar um modelo
ollama run llama3

# Listar modelos instalados
ollama list

# Ver modelos em execução
ollama ps

# Remover um modelo
ollama rm llama3

# Iniciar o servidor
ollama serve
```
